```python
from pathlib import Path
from visual_search_engine import SimilaritySearchEngine

image = Path("/home/anandas/test_images/Datewise/06/19/2021/IMG_20210619_165914.jpg")
engine = SimilaritySearchEngine(
    base_folder='/home/anandas/test_images',
    qdrant_url="http://localhost:6333",
)

embedding = engine.get_embedding(image)
results = engine.search(image, limit=20)
print(results)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load the image
img = mpimg.imread(image)

# Show the image
plt.imshow(img)
plt.axis('off')  # Hide axes
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import math

# Assume you already have a list of image paths
# Example:
# image_paths = [Path("/path/to/img1.jpg"), Path("/path/to/img2.png"), ...]

# Number of images per row
images_per_row = 5
n_images = len(results)
n_rows = math.ceil(n_images / images_per_row)

# Create figure
fig, axes = plt.subplots(n_rows, images_per_row, figsize=(20, 4 * n_rows))

# Flatten axes for easy iteration
axes = axes.flatten() if n_rows > 1 else axes

# Display images
for i, ax in enumerate(axes):
    if i < n_images:
        img = mpimg.imread(results[i]['filename'])
        ax.imshow(img)
        ax.set_title(f"score: {results[i]['score']}", fontsize=9)
        ax.axis('off')
       
        caption_text = results[i]['filename'].resolve().relative_to('/home/anandas/test_images')
        ax.text(0.5, -0.15, caption_text, transform=ax.transAxes,
            ha='center', va='top', fontsize=10, color='gray')

    else:
        ax.axis('off')  # Hide unused subplots

plt.tight_layout()
plt.show()
```
